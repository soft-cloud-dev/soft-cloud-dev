SoftCloud integrates Web Awesome via **authentic Lit SSR Declarative Shadow DOM (DSD)**. Components are rendered statically on the server during the build, minimizing cumulative layout shift (Page CLS <= 0.1, Component CLS <= 0.01) and hydrating seamlessly in the browser. Authors can create components either programmatically in Python notebooks using the `softcloud_ui.UI` API or declaratively in MyST Markdown using native directives and roles.

## 1. Choosing Components by User Intent (`choosing-components`)

When designing interactive technical documentation or laboratory notes, choose components based on **user intent** rather than visual styling alone:

| User Intent | Recommended Component | Allowed Variants / Options | Primary Use Case |
|---|---|---|---|
| **System Status & Feedback** | `<wa-callout>` | `brand`, `neutral`, `success`, `warning`, `danger` | Operational status, warnings, configuration caveats, test outcomes |
| **Command & Navigation** | `<wa-button>` | `brand`, `neutral`, `success`, `warning`, `danger` | Terminal actions, documentation links, script triggers, downloads |
| **Structured Surface / Entity** | `<wa-card>` | `outlined`, `filled`, `accent`, `filled-outlined`, `plain` | Server node specs, service dashboards, media previews, metric summaries |
| **Semantic Iconography** | `<wa-icon>` / Font Awesome | `solid`, `regular`, `brands` (FA 6.5.2) | Infrastructure icons, code symbols, security badges, brand logos |

### Interface Parity: Python API vs. MyST Directives
Authors can author the exact same component using either Python or MyST Markdown:

| Component | Python Notebook API | MyST Markdown Syntax |
|---|---|---|
| **Callout** | `UI.callout_text(text,`<br/>`&nbsp;&nbsp;title=..., variant=...)` | ````{wa-callout} Title`<br/>`:variant: brand`<br/>`Content...`<br/>```` |
| **Button** | `UI.button(label,`<br/>`&nbsp;&nbsp;href=..., variant=...)` | ````{wa-button} https://link`<br/>`:variant: brand`<br/>`Button Label`<br/>```` |
| **Card** | `UI.card_text(text,`<br/>`&nbsp;&nbsp;header=..., appearance=...)` | ````{wa-card} Card Header`<br/>`:appearance: filled-outlined`<br/>`Content...`<br/>```` |
| **Icon** | `UI.icon(name,`<br/>`&nbsp;&nbsp;family=..., variant=...)` | `{fa}` `solid server` or<br/>`{icon}` `rocket` |

## 2. Web Awesome Core Component Reference (`components/`)

SoftCloud provides first-class SSR support for four core Web Awesome custom elements. Below are their specifications, allowed attributes, and interactive showcases.

### 2.1 Buttons (`<wa-button>`)
Buttons trigger actions or navigate to resources. They support distinct semantic variants, surface appearances, sizes, and shape modifications:

- **Variants**: `brand` (primary theme action), `neutral` (secondary action), `success` (confirmation/positive action), `warning` (cautionary action), `danger` (destructive action).
- **Appearances**: `filled` (high contrast, default), `accent`, `outlined` (subtle border), `filled-outlined`, `plain` (ghost/flat button).
- **Sizes**: `xs`, `s`, `m` (medium / default), `l`, `xl`. Compact size tokens ensure zero runtime console deprecations.
- **Shapes & Modifiers**: `pill` (fully rounded caps), `circle` (equal aspect icon button), `disabled` (inaccessible state), `loading` (spinner state), `caret` (dropdown indicator).
- **Navigation**: `href` (validated link URL) and `target` (`_blank`, `_self`; automatically enforces `rel="noopener noreferrer"`).

```{wa-button} https://my.softcloud.dev/
:variant: brand
:appearance: filled
:size: m
Documentation Home Portal
```

`````python
# Button variants showcase
variants = ["brand", "neutral", "success", "warning", "danger"]
for var in variants:
    display(UI.button(f"{var.capitalize()} Action", variant=var, size="m"))
`````

`````python
# Button sizes, appearances, and pill modifiers
display(UI.button("Small Outlined", variant="brand", appearance="outlined", size="s"))
display(UI.button("Medium Filled Pill", variant="brand", pill=True, size="m"))
display(UI.button("Large Success Action", variant="success", size="l"))
display(UI.button("External Documentation", href="https://my.softcloud.dev/", variant="neutral", target="_blank"))
`````

### 2.2 Callouts (`<wa-callout>`)
Callouts highlight important notes, warnings, prerequisites, and status updates inline within prose or computational workflows:

- **Variants**:
  - `brand`: General announcements, feature updates, architectural highlights.
  - `neutral`: General notes, contextual background, tips.

  - `warning`: Deprecations, prerequisites, resource constraints.
  - `danger`: Critical errors, failure modes, security warnings.
- **Appearances**: `accent`, `filled`, `outlined`, `filled-outlined`, `plain`.
- **Title Slot**: Accessible header text announcing the callout purpose (rendered in default content slot).

```{wa-callout} Operational Integrity Notice

All node telemetry, security policies, and health probes pass continuously with minimal layout shift.
```

### 2.3 Cards (`<wa-card>`)

## 4. Design Tokens & CSS Utilities Reference (`tokens/`, `utilities/`)

Web Awesome is built on a design token architecture. CSS custom properties allow consistent spacing, color grading, shadows, and responsive typography across all components.

### 4.1 Design Token Reference

| Token Family | CSS Variable Pattern | Example Value | Description |
|---|---|---|---|
| **Brand Colors** | `--wa-color-brand-{05-95}` | `var(--wa-color-brand-60)` | Primary accent hue for interactive elements and brand accents |
| **Neutral Grays** | `--wa-color-gray-{05-95}` | `var(--wa-color-gray-20)` | Surface backgrounds, borders, and structural dividers |
| **Semantic Colors** | `--wa-color-{success\|warning\|danger}-{60}` | `var(--wa-color-success-60)` | State indicators for health, warnings, and errors |
| **Text Contrast** | `--wa-color-text-{normal\|quiet\|loud}` | `var(--wa-color-text-normal)` | Accessible text contrast hierarchy |
| **Spacing Scale** | `--wa-space-{3xs\|2xs\|xs\|s\|m\|l\|xl\|2xl\|3xl}` | `var(--wa-space-m)` (1.0rem) | Consistent margin, padding, and layout gap scale |
| **Border Radius** | `--wa-border-radius-{xs\|s\|m\|l\|full}` | `var(--wa-border-radius-m)` (0.5rem) | Container corner rounding |
| **Elevation / Shadow** | `--wa-shadow-{xs\|sm\|md\|lg\|xl}` | `var(--wa-shadow-md)` | Depth layers for cards, menus, and overlays |

## 5. Usage Patterns, Themes & Architecture (`usage`, `themes`)

### 5.1 Lit SSR & Declarative Shadow DOM (DSD) Lifecycle
In SoftCloud, Web Awesome web components follow a modern three-stage lifecycle:

1. **Build-Time Transformation (`scripts/lit-ssr-transform.mjs`)**:
   During site compilation, HTML placeholders produced by MyST directives and Jupyter notebook outputs are parsed using `parse5`. The component render model generates custom elements and attaches `<template shadowroot="open" shadowrootmode="open">` containing the server-rendered template.
2. **Instant Paint (Page CLS <= 0.1, Component CLS <= 0.01)**:
   The browser immediately renders the component internals via native Declarative Shadow DOM before client-side scripts download or execute. There is no cumulative layout shift or flash of unstyled content.
3. **Seamless Hydration (`webawesome.ssr-loader.js`)**:
   The SSR autoloader adopts the existing shadow root without destroying or re-rendering the DOM tree, attaching event listeners and reactivity in place.

### 5.2 Theme Synchronization
SoftCloud's `softcloud-theme.js` automatically coordinates theme state:
- Matches MyST book-theme light and dark mode toggles.
- Applies `.wa-dark` / `.wa-light` classes to the `<html>` root element.
- Design tokens automatically remap contrast ratios, backgrounds, and text shades for optimal readability.

```{wa-callout} Theme Coordination Status
:variant: brand
:appearance: filled-outlined
Theme synchronization operates synchronously via softcloud-theme.js without client hydration mismatch.
```
