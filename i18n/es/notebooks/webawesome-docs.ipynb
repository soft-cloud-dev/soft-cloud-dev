## 1. Choosing the Right Component (`choosing-components`)

When selecting a component, start from **user intent** rather than component name:

### Decision Matrix
| User Intent | Component Choice | Key Considerations |
|---|---|---|
| **Pick one (2–5 options)** | `<wa-radio-group>` | All options visible simultaneously. |
| **Pick one (many options)** | `<wa-select>` with `<wa-option>` | Compact dropdown selector. *(Do not use `<wa-dropdown>` for value selection!)* |
| **Pick many** | `<wa-select multiple>` or `<wa-checkbox-group>` | Multi-selection from a set. |
| **Instant toggle** | `<wa-switch>` | Takes effect immediately (e.g. notifications on/off). |
| **Form boolean** | `<wa-checkbox>` | Submitted with form payload later. |
| **Trigger action** | `<wa-button>` | Primary action uses `variant="brand"`. |
| **Action menu** | `<wa-dropdown>` | Menu of commands/actions (not for picking form values). |
| **Feedback / status** | `<wa-callout>`, `<wa-badge>`, `<wa-toast>` | Persistent inline message or compact badge. |
| **Data display** | `<wa-card>`, `<wa-avatar>`, `<wa-format-date>` | Structured container with slots for header, media, and footer. |
| **Overlay** | `<wa-dialog>`, `<wa-drawer>`, `<wa-popover>` | Modal dialogs and flyout drawers. |

## 3. Framework Integration (`frameworks/`)

Web Awesome custom elements work across all major modern JavaScript frameworks:

### React
Use Web Awesome with standard JSX or Lit React wrappers (`@lit/react`):
```tsx
import { createComponent } from '@lit/react';
import React from 'react';
import { WaButton } from '@awesome.me/webawesome/dist/components/button/button.js';

export const Button = createComponent({
  tagName: 'wa-button',
  elementClass: WaButton,
  react: React,
  events: { onWaChange: 'wa-change' },
});
```

### Vue
Configure `isCustomElement` in `vite.config.ts`:
```ts
vue({
  template: {
    compilerOptions: {
      isCustomElement: (tag) => tag.startsWith('wa-')
    }
  }
})
```

### Angular
Add `CUSTOM_ELEMENTS_SCHEMA` to your module/component:
```ts
import { CUSTOM_ELEMENTS_SCHEMA } from '@angular/core';

@Component({
  schemas: [CUSTOM_ELEMENTS_SCHEMA]
})
```

### Svelte
Custom elements are first-class in Svelte with direct binding:
```svelte
<wa-button on:click={handleClick} variant="brand">Submit</wa-button>
```

## 4. Utilities (`utilities/`)

Web Awesome provides utility CSS classes in `styles/webawesome.css`:

- **Prose**: `.wa-prose` provides styled typography for markdown articles.
- **Rounding**: `.wa-rounded-none`, `.wa-rounded-sm`, `.wa-rounded-md`, `.wa-rounded-lg`, `.wa-rounded-full`.
- **Text Styling**: `.wa-text-quiet` (secondary), `.wa-text-normal`, `.wa-text-loud` (bold/high-contrast).

````python
# Demonstrating utility styling with TrustedHTML
html_markup = """
<div style="padding: 1rem; border: 1px dashed var(--wa-color-gray-40); border-radius: var(--wa-border-radius-m);">
  <span class="wa-text-loud">Bold Utility Text</span> &mdash; 
  <span class="wa-text-quiet">Quiet Utility Subtitle</span>
  <span class="wa-visually-hidden">Screen reader only announcement</span>
</div>
"""
UI.card_trusted_html(TrustedHTML(html_markup), header="CSS Utility Showcase")
````

## 5. Design Tokens (`tokens/`)

Design tokens define the visual foundation of Web Awesome:

- **Color Scale**: `--wa-color-{hue}-{tint}` from `05` to `95` (e.g. `--wa-color-brand-60`, `--wa-color-gray-20`).
- **Spacing Scale**: `--wa-space-3xs` through `--wa-space-3xl` (`--wa-space-m` = `1rem`).
- **Border Radius**: `--wa-border-radius-s`, `--wa-border-radius-m`, `--wa-border-radius-l`, `--wa-border-radius-full`.
- **Shadows**: `--wa-shadow-sm`, `--wa-shadow-md`, `--wa-shadow-lg`, `--wa-shadow-xl`.
- **Focus**: `--wa-focus-ring-width: 2px`, `--wa-focus-ring-color: var(--wa-color-primary-60)`.

````python
# Inspecting available color and spacing tokens
tokens_summary = {
    "Brand Primary": "var(--wa-color-brand-60)",
    "Text Normal": "var(--wa-color-text-normal)",
    "Panel Radius": "var(--wa-panel-border-radius)",
    "Default Spacing": "var(--wa-space-m)"
}
for k, v in tokens_summary.items():
    print(f"Token: {k:16} => {v}")
````

## 6. Installation & Loading (`installation`)

Web Awesome supports three primary distribution modes:

1. **Self-Hosted SSR (Recommended in SoftCloud)**:
   Assets are staged to `vendor/webawesome` and transformed statically using `@lit-labs/ssr` Declarative Shadow DOM.
2. **npm Package**:
   ```bash
   npm install @awesome.me/webawesome@3.14.0
   ```
3. **CDN Loading**:
   ```html
   <link rel="stylesheet" href="https://ka-f.webawesome.com/webawesome@3.14.0/styles/webawesome.css" />
   <script type="module" src="https://ka-f.webawesome.com/webawesome@3.14.0/webawesome.loader.js"></script>
   ```

## 7. Usage Patterns & SSR (`usage`)

Key rules when working with Web Awesome:
- **No self-closing tags**: Web Components are custom elements and require closing tags: `<wa-button>...</wa-button>`.
- **Slots**: Use `slot="start"`, `slot="end"`, `slot="header"`, `slot="footer"` to position children inside shadow roots.
- **Declarative Shadow DOM**: The SSR engine attaches `<template shadowroot="open" shadowrootmode="open">` to custom element hosts, ensuring instant rendering before JavaScript hydration.
- **SSR Autoloader**: `vendor/webawesome/webawesome.ssr-loader.js` hydrates DSD components without double-rendering.

````python
# Demonstrating slot composition in Python
card_content = """
<p>This card demonstrates named slot composition:</p>
<ul>
  <li><strong>Header slot:</strong> Card Title</li>
  <li><strong>Default slot:</strong> Body paragraph</li>
  <li><strong>Action:</strong> Embedded interactive button</li>
</ul>
"""
UI.card_trusted_html(
    TrustedHTML(card_content),
    header="Slot Composition Pattern",
    appearance="elevated"
)
````

Web Awesome features a tokenized theme engine:
